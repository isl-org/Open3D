```python
import open3d as o3d
from open3d._build_config import _build_config

if not _build_config["BUILD_JUPYTER_EXTENSION"]:
    raise RuntimeError(
        "This Open3D package was built without Jupyter support. "
        "Install an official wheel or rebuild with "
        "-DBUILD_WEBRTC=ON -DBUILD_JUPYTER_EXTENSION=ON.")

from open3d.web_visualizer import draw
```

```python
cube_red = o3d.geometry.TriangleMesh.create_box(1, 2, 4)
cube_red.compute_vertex_normals()
cube_red.paint_uniform_color((1.0, 0.0, 0.0))
draw(cube_red)
```

```python
def show_controls(visualizer):
    visualizer.show_message_box(
        "Open3D Controls",
        "Use the mouse to orbit, pan, and zoom. Open Help > Show Controls... "
        "for the complete control reference.")


draw(cube_red, actions=[("Show Controls", show_controls)])
```
